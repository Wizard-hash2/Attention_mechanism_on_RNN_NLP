```python
import torch

sentence = torch.tensor([7,2,8,5,6,8,4])

sentence

torch.manual_seed(1)
embed = torch.nn.Embedding(10,16) #10 is the dictionary size, 16 is the embedding size
embedded_sentence = embed(sentence).detach()
embedded_sentence.shape

#now matrix multication
omega_mat = embedded_sentence.matmul(embedded_sentence.T)

#now computing attention weights using the pytorch softmax 

import torch.nn.functional as F
attention_weights = F.softmax(omega_mat, dim =1)

attention_weights.shape

#now finally computing z, the context aware embedding vectord

#by summing total attention * x
x_2 = embedded_sentence[1,:]
context_vec_2 = torch.zeros(x_2.shape)
for j in range(7):
    x_j = embedded_sentence[j, :]
    context_vec_2 += attention_weights[1, j] * x_j 
context_vec_2

# or instead of the above for we can take advantage of the matmul
context_vec = torch.matmul(attention_weights, embedded_sentence)
context_vec

# torch.allclose(context_vec_2, context_vec[1]) .. similarity
```
